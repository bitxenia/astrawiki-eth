```python
import matplotlib.pyplot as plt
import numpy as np
import json

# Load data from the uploaded JSON files
with open("./metrics/newArticle_limit_gasUsed_1748563262744.json") as f:
    gas_used_raw = json.load(f)

with open("./metrics/newArticle_limit_time_1748563262742.json") as f:
    times = json.load(f)

# Calculate averages of every 10 lines for gas_used
gas_used = [np.mean(gas_used_raw[i:i+10]) for i in range(0, len(gas_used_raw), 10)]

# Assume each entry corresponds to an article in 1000 byte increments
increments = np.arange(1000, 1000 * (len(gas_used) + 1), 1000)

# Plot gas used vs article size
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(increments, gas_used, marker="o", color="blue")
plt.title("Gas Usado vs Tamaño del Articulo")
plt.xlabel("Tamaño del Articulo (bytes)")
plt.ylabel("Gas Usado (ETH)")
plt.grid(True)

# Plot time taken vs article size
plt.subplot(1, 2, 2)
plt.plot(increments, times, marker="x", color="green")
plt.title("Tiempo vs Tamaño del Articulo")
plt.xlabel("Tamaño del Articulo (bytes)")
plt.ylabel("Tiempo (ms)")
plt.grid(True)

plt.tight_layout()
plt.show()
```
